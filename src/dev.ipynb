```python
from stung.bumble import *
from stung.utils import *
```

```python
genome_fp = "/ccb/salz4-3/hji20/hg002-q100-annotation/data/hg002v1.1.fasta"
ann_fps = [
    "/ccb/salz4-3/hji20/hg002-q100-annotation/synteny/sub_annots/chr1_mat.gff",
    "/ccb/salz4-3/hji20/hg002-q100-annotation/synteny/sub_annots/chr1_pat.gff"
]
temp_dir = "/ccb/salz4-3/hji20/stung/temp"
out_dir = "/ccb/salz4-3/hji20/stung/out"
```

```python
bmbl = Bumble(
    genome_fp = genome_fp,
    ann_fps = ann_fps,
    temp_dir = temp_dir,
    out_dir = out_dir
)
```

```python
mat = bmbl.build_2d_matrix()
```

```python
import plotly.graph_objects as go
```

```python
custom_colorscale = [
            [0.0, "white"],
            [0.333, "white"],
            [0.334, "blue"],
            [0.666, "blue"],
            [0.667, "red"],
            [1.0, "red"]
        ]
fig = go.Figure(data=go.Heatmap(
            z=mat,
            colorscale=custom_colorscale,
            showscale=False,  # hide colorbar
            hovertemplate="x: %{x}<br>y: %{y}<br>value: %{z}<extra></extra>",
        ))

fig.update_layout(
    dragmode='zoom',
    xaxis=dict(scaleanchor='y'),
)

fig.show()
```
